```python
import pandas as pd
import numpy as np
import random

# Seed for reproducibility
np.random.seed(42)
random.seed(42)

# Generate Employee IDs
employee_ids = [f"E{i:04d}" for i in range(1, 1001)]

# Likert-scale generator
def likert_score(scale_len, dist):
    return np.random.choice(range(1, scale_len + 1), p=dist)

# Initialize dataset
wellness_data = []

for emp_id in employee_ids:
    burnout = likert_score(7, [0.05, 0.10, 0.20, 0.25, 0.20, 0.15, 0.05])  # SHRM, Gallup, Zhong et al.
    wellbeing = likert_score(7, [0.02, 0.05, 0.10, 0.25, 0.30, 0.20, 0.08])  # Gallup, Reeja et al.
    anxiety = likert_score(5, [0.10, 0.15, 0.30, 0.25, 0.20])  # Reeja et al.
    depression = likert_score(5, [0.10, 0.20, 0.30, 0.25, 0.15])  # Reeja et al.
    stress = likert_score(5, [0.05, 0.10, 0.25, 0.30, 0.30])  # Gallup

    # Mental health leave – linked to burnout and stress
    mental_days = np.random.poisson(lam=1 if burnout >= 5 or stress >= 4 else 0.3)

    # Absenteeism – influenced by anxiety or depression levels
    absenteeism = np.random.poisson(lam=2 if anxiety >= 4 or depression >= 4 else 0.5)

    # Access to counselor – used in literature for moderating stress and emotional regulation (Gallup, SHRM)
    access_to_counseling = np.random.choice(["Yes", "No"], p=[0.35, 0.65])  # ~35% of orgs offer it (SHRM)

    # Add some slight noise
    mental_days = np.random.poisson(lam=(1.2 if burnout >= 5 or stress >= 4 else 0.4))
    absenteeism = np.random.poisson(lam=(2.2 if anxiety >= 4 or depression >= 4 else 0.6))

    wellness_data.append([
        emp_id,
        burnout,
        wellbeing,
        anxiety,
        depression,
        stress,
        mental_days,
        absenteeism,
        access_to_counseling
    ])

# Define column names
columns = [
    "Employee_ID",
    "Burnout_Score",                        # SHRM, Gallup
    "Wellbeing_Score",                      # Gallup, Reeja et al.
    "Anxiety_Level",                        # Reeja et al.
    "Depression_Level",                     # Reeja et al.
    "Stress_Level",                         # Gallup
    "Mental_Health_Days_Taken",             # Gallup
    "Absenteeism_Due_To_Mental_Health",     # Gallup
    "Access_To_Counseling"                  # SHRM
]

# Create DataFrame
wellness_df = pd.DataFrame(wellness_data, columns=columns)

# Save to Excel
wellness_df.to_excel("Employee_Wellness.xlsx", index=False)
print("Employee_Wellness.")
```
