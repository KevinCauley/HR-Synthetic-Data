```python
import pandas as pd
import numpy as np

# Set seed for reproducibility
np.random.seed(42)

# Generate consistent Employee IDs
employee_ids = [f"E{i:04d}" for i in range(1, 1001)]

# Likert scoring helper
def likert_score(scale, dist):
    return np.random.choice(range(1, scale + 1), p=dist)

# Compensation & Output data
comp_data = []

for emp_id in employee_ids:
    # Training: number of hours (Fallucchi et al.)
    training_hours = max(0, int(np.random.normal(loc=50, scale=20)))

    # Performance: ordinal 1–5 (Al Akasheh et al.)
    performance = np.random.choice([1, 2, 3, 4, 5], p=[0.05, 0.10, 0.35, 0.35, 0.15])

    # Productivity Score: self-reported or manager-rated (Gallup-style)
    productivity = np.random.choice([1, 2, 3, 4, 5], p=[0.05, 0.10, 0.30, 0.35, 0.20])

    # Extra Effort: how often employee goes beyond minimum (Gallup, SHRM, Zhong)
    effort = np.random.choice([1, 2, 3, 4, 5], p=[0.05, 0.10, 0.30, 0.35, 0.20])

    # Engagement: Gallup Q12 style
    engagement = likert_score(5, [0.10, 0.20, 0.30, 0.25, 0.15])

    # Organizational Commitment (Zhong et al., Arar et al.)
    commitment = likert_score(5, [0.05, 0.10, 0.25, 0.35, 0.25])

    comp_data.append([
        emp_id,
        training_hours,
        performance,
        productivity,
        effort,
        engagement,
        commitment
    ])

# Create DataFrame
columns = [
    "Employee_ID",
    "Actual_Training_Hours",
    "Performance_Score",
    "Productivity_Score",
    "Extra_Effort_Score",
    "Engagement_Level",
    "Organizational_Commitment"
]

comp_df = pd.DataFrame(comp_data, columns=columns)

# Save to Excel
comp_df.to_excel("Employee_Metrics.xlsx", index=False)
print("Employee_Metrics.xlsx saved.")
```
