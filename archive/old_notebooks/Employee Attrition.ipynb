```python
import pandas as pd
import numpy as np

# Reproducibility
np.random.seed(42)

# Generate Employee IDs
employee_ids = [f"E{i:04d}" for i in range(1, 1001)]

# Function to simulate binary outcomes
def binary_outcome(prob_yes):
    return np.random.choice(["Yes", "No"], p=[prob_yes, 1 - prob_yes])

# Function for Likert scores
def likert_score(dist):
    return np.random.choice(range(1, len(dist)+1), p=dist)

# Function for low-skewed numeric distribution
def count_score(probabilities, values):
    return np.random.choice(values, p=probabilities)

# Initialize dataset
attrition_data = []

for emp_id in employee_ids:
    turnover_intention = binary_outcome(0.25)  # Gallup, Zhong et al.
    voluntary_resignation = binary_outcome(0.15)  # SHRM benchmark
    involuntary_termination = binary_outcome(0.05)  # Industry average
    quiet_quitting = binary_outcome(0.22)  # Fallucchi, Zhong et al.

    # Likert-style 1–5 scale: Psychological Contract Violation
    employer_promise_broken = likert_score([0.10, 0.20, 0.30, 0.25, 0.15])  # Zhong et al., Wells et al.

    # Internal job applications (typically 0–3 per year)
    internal_job_applications = count_score([0.60, 0.25, 0.10, 0.05], [0, 1, 2, 3])  # SHRM mobility stats

    # Recent promotion: 0 = No, 1 = Yes (rare event)
    promotion_last_12_months = np.random.choice([0, 1], p=[0.85, 0.15])  # Reflects lower promotion frequency
    
    # Voluntary Resignation dependent on Turnover Intention
    if turnover_intention == "Yes":
        voluntary_resignation = binary_outcome(0.40)
    else:
        voluntary_resignation = binary_outcome(0.10)
        
    attrition_data.append([
        emp_id,
        turnover_intention,
        voluntary_resignation,
        involuntary_termination,
        quiet_quitting,
        employer_promise_broken,
        internal_job_applications,
        promotion_last_12_months
    ])

# Define column names
columns = [
    "Employee_ID",
    "Turnover_Intention",            # Predictor of actual resignation (Zhong et al., Gallup)
    "Voluntary_Resignation",         # Primary attrition outcome (target)
    "Involuntary_Termination",       # Supplementary attrition context
    "Quiet_Quitting",                # Disengagement signal (Zhong et al., Fallucchi)
    "Employer_Promise_Broken",       # Contract breach perception (Wells et al.)
    "Internal_Job_Applications",     # Job search behavior within company
    "Promotion_Last_12_Months"       # Career advancement experience
]

# Create DataFrame
attrition_df = pd.DataFrame(attrition_data, columns=columns)

# Save to Excel
attrition_df.to_excel("Employee_Attrition.xlsx", index=False)
print("Employee_Attrition.")
```
