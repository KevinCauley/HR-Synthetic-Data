```python
import pandas as pd
import numpy as np
import random

# Seed for reproducibility
np.random.seed(42)
random.seed(42)

# Generate IDs
employee_ids = [f"E{i:04d}" for i in range(1, 1001)]

# Time and Surveillance Options
time_tracking_methods = [
    "Clock-in/out", "Timesheets", "Keystroke Monitoring", "Activity Timer", "None"
]
surveillance_methods = [
    "Screenshot Capturing", "Webcam Monitoring", "Screen Recording",
    "Email Scanning", "Browser History Logging", "Application Usage Tracking", "None"
]

# Likert Helper
def likert_score(levels=5, dist=None):
    if dist is None:
        dist = [1/levels] * levels
    return np.random.choice(range(1, levels+1), p=dist)

# Initialize data list
monitoring_data = []

# Loop for data generation
for emp_id in employee_ids:
    tracking = np.random.choice(time_tracking_methods, p=[
        0.20,  # Clock-in/out
        0.25,  # Timesheets
        0.15,  # Keystroke Monitoring
        0.25,  # Activity Timer
        0.15   # None
    ])  # Cisco, SHRM

    surveillance = np.random.choice(surveillance_methods, p=[
        0.15,  # Screenshot Capturing
        0.10,  # Webcam Monitoring
        0.15,  # Screen Recording
        0.20,  # Email Scanning
        0.20,  # Browser Logging
        0.15,  # App Usage
        0.05   # None
    ])  # Wells et al., Forbes

    total_monitoring_tools = min(np.random.poisson(2), 6)  # Most IT workers 1–3 (Wells et al.)

    monitoring_frequency = likert_score(5, [0.10, 0.20, 0.35, 0.25, 0.10])  # Gallup

    violations = np.random.poisson(lam=1 if monitoring_frequency > 3 else 0.3)  # Cisco

    monitoring_data.append([
        emp_id, tracking, surveillance,
        total_monitoring_tools, monitoring_frequency, violations
    ])

# Column labels
columns = [
    "Employee_ID",
    "Time_Tracking_Method_Used",
    "Surveillance_Method_Used",
    "Total_Monitoring_Tools",
    "Monitoring_Frequency",
    "Violations_Logged_By_Monitoring_Software"
]

# Create and export
monitoring_df = pd.DataFrame(monitoring_data, columns=columns)
monitoring_df.to_excel("Employee_Monitoring.xlsx", index=False)
print("Employee_Monitoring.xlsx saved.")
```
