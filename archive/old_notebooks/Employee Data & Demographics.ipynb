```python
import pandas as pd
import numpy as np
import random
from faker import Faker

# Initialize generators and seeds
fake = Faker()
Faker.seed(42)
np.random.seed(42)
random.seed(42)

# Constants
num_employees = 1000

# Define departments by occupation mapping
occupation_department_map = {
    "Software Developer": "Applications",
    "Information Security Analyst": "Security",
    "Computer Systems Analyst": "Infrastructure",
    "Computer Support Specialist": "Support",
    "Computer Network Architect": "Infrastructure",
    "Computer Programmer": "Applications",
    "Web Developer": "Web",
    "Computer and Information Systems Manager": "Management",
    "Network and Computer Systems Administrator": "Infrastructure",
    "Software Quality Assurance Analyst and Tester": "QA",
    "Web and Digital Interface Designer": "Web"
}

# Occupation-specific salary percentiles (approximate medians)
occupation_salary_map = {
    "Software Developer": 127260,
    "Information Security Analyst": 120360,
    "Computer Systems Analyst": 107360,
    "Computer Support Specialist": 60740,
    "Computer Network Architect": 126900,
    "Computer Programmer": 96400,
    "Web Developer": 87820,
    "Computer and Information Systems Manager": 169510,
    "Network and Computer Systems Administrator": 97200,
    "Software Quality Assurance Analyst and Tester": 99620,
    "Web and Digital Interface Designer": 86610
}

seniority_levels = ["Entry", "Junior", "Mid", "Senior", "Lead"]
age_groups = ["16-19", "20-24", "25-34", "35-44", "45-54", "55-64", "65+"]
genders = ["Male", "Female"]
races = ["White", "Black or African American", "Asian", "Other"]
ethnicities = ["Hispanic", "Non-Hispanic"]

# Likert helper
def likert_score(p=[0.10, 0.20, 0.30, 0.25, 0.15]):
    return np.random.choice([1, 2, 3, 4, 5], p=p)

# Generate employee records
employee_data = []
for i in range(1, num_employees + 1):
    emp_id = f"E{i:04d}"
    first_name = fake.first_name()
    last_name = fake.last_name()

    occupation = random.choice(list(occupation_department_map.keys()))
    department = occupation_department_map[occupation]

    # BLS tenure distribution for IT occupations
    tenure_range = np.random.choice(["0-1", "1-3", "3-5", "5-10", "10+"], p=[0.30, 0.40, 0.15, 0.10, 0.05])
    if tenure_range == "0-1":
        tenure = round(np.random.uniform(0.1, 1.0), 1)
        salary_percentile = 0.10
        seniority = "Entry"
    elif tenure_range == "1-3":
        tenure = round(np.random.uniform(1.0, 3.0), 1)
        salary_percentile = 0.25
        seniority = "Junior"
    elif tenure_range == "3-5":
        tenure = round(np.random.uniform(3.0, 5.0), 1)
        salary_percentile = 0.50
        seniority = "Mid"
    elif tenure_range == "5-10":
        tenure = round(np.random.uniform(5.0, 10.0), 1)
        salary_percentile = 0.75
        seniority = "Senior"
    else:
        tenure = round(np.random.uniform(10.0, 20.0), 1)
        salary_percentile = 0.90
        seniority = "Lead"

    # Base salary based on occupation and tenure-influenced percentile
    base_salary = occupation_salary_map[occupation]
    salary_multiplier = np.random.normal(loc=salary_percentile, scale=0.05)
    salary_multiplier = min(max(salary_multiplier, 0.05), 1.0)
    average_salary = int(base_salary * salary_multiplier)

    # Work arrangement defaults
    age_group = np.random.choice(age_groups, p=[0.01, 0.05, 0.25, 0.25, 0.20, 0.20, 0.04])
    gender = np.random.choice(genders, p=[0.74, 0.26])
    race = np.random.choice(races, p=[0.64, 0.10, 0.23, 0.03])
    ethnicity = np.random.choice(ethnicities, p=[0.08, 0.92])

    # Work arrangement logic (Gallup + McKinsey)
    if gender == "Female" and age_group in ["35-44", "45-54", "55-64"]:
        work_arrangement = np.random.choice(["Hybrid", "Remote", "On-site"], p=[0.55, 0.35, 0.10])
    elif gender == "Male" and age_group in ["35-44", "45-54", "55-64"]:
        work_arrangement = np.random.choice(["Hybrid", "Remote", "On-site"], p=[0.50, 0.30, 0.20])
    else:
        work_arrangement = np.random.choice(["Hybrid", "Remote", "On-site"], p=[0.50, 0.25, 0.25])

    employee_data.append([
        emp_id, first_name, last_name, occupation, department,
        average_salary, tenure, seniority, work_arrangement,
        age_group, gender, race, ethnicity
    ])

# Define column names in requested order
columns = [
    "Employee_ID", "First_Name", "Last_Name", "Occupation", "Department",
    "Average_Salary", "Tenure_Years", "Seniority", "Work_Arrangement",
    "Age_Group", "Gender", "Race", "Ethnicity"
]

# Create DataFrame
employee_df = pd.DataFrame(employee_data, columns=columns)

# Save to Excel
employee_df.to_excel("Employee_Data_and_Demographics.xlsx", index=False)
print("Employee_Data_and_Demographics saved as 'Employee_Data_and_Demographics.xlsx'")
```
