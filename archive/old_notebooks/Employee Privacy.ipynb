```python
import pandas as pd
import numpy as np

# Set seed for reproducibility
np.random.seed(42)

# Generate 1000 Employee IDs
employee_ids = [f"E{i:04d}" for i in range(1, 1001)]

# Likert scale generator
def likert_score(dist):
    return np.random.choice(range(1, len(dist) + 1), p=dist)

# Initialize list for data
privacy_data = []

for emp_id in employee_ids:
    monitoring_transparency = likert_score([0.10, 0.15, 0.25, 0.30, 0.20])
    # Justification: Gallup & Ravid et al. show moderate transparency in most organizations

    privacy_concern = likert_score([0.05, 0.10, 0.25, 0.30, 0.30])
    # Employees express strong concern about how data is used and stored (Wells et al.)

    perceived_intrusiveness = likert_score([0.05, 0.15, 0.25, 0.30, 0.25])
    # Many employees report discomfort, especially in IT roles with surveillance (Wells et al.)

    trust_in_monitoring = likert_score([0.10, 0.15, 0.25, 0.30, 0.20])
    # Trust is a mixed sentiment — higher if transparency is high (Gallup, Ravid)

    # Complaints rise if perceived intrusiveness is 4 or 5
    complaints = np.random.poisson(lam=0.7 if perceived_intrusiveness > 3 else 0.2)

    privacy_data.append([
        emp_id,
        monitoring_transparency,
        privacy_concern,
        perceived_intrusiveness,
        trust_in_monitoring,
        int(complaints)
    ])

# Create DataFrame
columns = [
    "Employee_ID",
    "Monitoring_Transparency",       # Ravid et al.
    "Privacy_Concern_Score",         # Wells et al.
    "Perceived_Intrusiveness",       # Wells et al.
    "Trust_In_Monitoring",           # Gallup, Ravid et al.
    "Privacy_Complaints_Score"       # Derived from intrusiveness
]

privacy_df = pd.DataFrame(privacy_data, columns=columns)

# Export to Excel
privacy_df.to_excel("Employee_Privacy.xlsx", index=False)
print("Employee_Privacy.xlsx saved.")
```
